```python
import pandas as pd
```

```python
# Pandas Series
# Sabse pehle tumhari list ko Series mein convert karte hain:
data = [10, 20, 30, 40, 50]

series = pd.Series(data)

print(series)
```

```python
# Custom Index
# Ab apna index khud set karte hain
data = [10, 20, 30, 40, 50]

series = pd.Series(
    data,
    index=["A", "B", "C", "D", "E"]
)

print(series)
```

```python
age = [19,19,18,15,19]
series = pd.Series(
    age,
    index=["Abdullah", "Bilal", "uman", "hashim", "Esham "]
)

print(series)
```

```python
# 1. List se DataFrame
data = [
    ["Ali", "Lahore", 5000],
    ["Ahmed", "Karachi", 7000],
    ["Hamza", "Islamabad", 6000]
]

df = pd.DataFrame(
    data,
    columns=["Customer", "City", "Sales"]
)

print(df)
```

```python
data = [
    ["Ali", "Lahore", 5000],
    ["Ahmed", "Karachi", 7000],
    
]

df = pd.DataFrame(
    data,
    columns=["Customer", "City", "Sales"]
)

print(df)
```

```python
# data → list
# columns → column names
# pd.DataFrame() → list ko DataFrame mein convert karta hai
data = [
    ["Ali", "Lahore", 5000],
    ["Ahmed", "Karachi", 7000],
    ["Hamza", "Islamabad", 6000]
]

df = pd.DataFrame(
    data,
    columns=["Customer", "City", "Sales"]
)

print(df)
```

```python
# Dictionary se DataFrame
data = {
    "Customer": ["Ali", "Ahmed", "Hamza"],
    "City": ["Lahore", "Karachi", "Islamabad"],
    "Sales": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

print(df)
```
