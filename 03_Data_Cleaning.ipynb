```python
import pandas as pd
df = pd.read_csv("Online Retail.csv")
```

```python
df.head()
```

```python
# null values check karne ka liye
df.isnull().sum()
```

```python
# yeh bataye ga dataset ma total kitne missing value hai
df.isnull().sum().sum()
```

```python
# Column-wise missing values check karne ka liye
df.isnull().sum()
```

```python
# Missing Values ki Percentage nikalne ka formula
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage
```

```python
missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data
```

```python
# Duplicate Rows Check karne ka liye
df.duplicated().sum()
```

```python
df[df.duplicated()].head()
```

```python
# Agar tum complete duplicate rows ka count aur percentage ek saath dekhna chaho:
duplicate_count = df.duplicated().sum()
duplicate_percentage = (duplicate_count / len(df)) * 100

print("Duplicate Rows:", duplicate_count)
print("Duplicate Percentage:", duplicate_percentage)
```

```python
# drop_duplicates()
before = len(df)
# Ab duplicates remove karo:
df = df.drop_duplicates()
after = len(df)
print("Before:", before)
print("After:", after)
print("Removed:", before - after)
df.duplicated().sum()
```

```python
# har column ka datatype check karne ka liye
df.dtypes
```

```python
# InvoiceDate ko datetime mein convert karna
# InvoiceDate abhi string hai. Date-related analysis ke liye ise proper datetime banana zaroori hai.
df["InvoiceDate"].head()
# ab datetime ma covert karne ka liye
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["InvoiceDate"].dtype
# Datetime banne ke baad hum future mein easily:
# year kaliye
df["InvoiceDate"].dt.year
# month ka liye
df["InvoiceDate"].dt.month
# day ka liye
df["InvoiceDate"].dt.day
# hour ka liye
df["InvoiceDate"].dt.hour
# day name ka liye
df["InvoiceDate"].dt.day_name()
```

```python
df["InvoiceDate"].head()
```

```python
# sab ko aik sath check karne  ka liye 
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["Hour"] = df["InvoiceDate"].dt.hour

df[["InvoiceDate", "Year", "Month", "Day", "Hour"]].head()
```

```python
# Invalid / Negative Values check.
# Online Retail data mein Quantity aur UnitPrice ko blindly clean nahi karna. Pehle check karenge ke negative values kitni hain.
# Negative Value
(df["Quantity"] < 0).sum()
# Unit Value
(df["UnitPrice"] < 0).sum()
# Zero unit Value
(df["UnitPrice"] == 0).sum()
df = df[df["Quantity"] > 0]
```

```python
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Negative UnitPrice:", (df["UnitPrice"] < 0).sum())
print("Zero UnitPrice:", (df["UnitPrice"] == 0).sum())
```

```python
# Negative UnitPrice ki rows
df[df["UnitPrice"] < 0]
df[df["UnitPrice"] == 0].head(10)
```

```python
# Isse humein saare zero + negative price transactions ek jagah mil jayenge.
df[df["UnitPrice"] <= 0][
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice", "CustomerID"]
]
```

```python
# is ma jo value 0 ka value hai wo pata lag jae gi
df[df["UnitPrice"] == 0][
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice", "CustomerID"]
].head(20)
```

```python
df[df["UnitPrice"] < 0][
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice", "CustomerID"]
].head(20)
```

```python
# CustomerID missing values
# Tumhare data mein CustomerID ki 135,080 missing values (~24.93%) hain.
# Pehle check karte hain ke missing CustomerID wali transactions kitni hain:
df[df["CustomerID"].isna()].head()
```

```python
# total missing value check karne ka liye 
df["CustomerID"].isna().sum()
```

```python
# ab descripton i total missing values
df[df["Description"].isna()].head()
```

```python
# ab description aur customer id ki dekhi
df[df["CustomerID"].isna()].head()
df[df["Description"].isna()].head()
```

```python
# Description missing rows remove
df = df.dropna(subset=["Description"])
df
```

```python
# ab verify kare ka koi missing value hai toh nhi 
df["Description"].isna().sum()
```

```python
# ab Customer ki missing check karke remove kare
df[df["CustomerID"].isna()].head()
```

```python
df = df.dropna(subset=["CustomerID"])
df
```

```python
# ab verify kare ka koi missing value hai toh nhi 
df["CustomerID"].isna().sum()
```

```python
# overall missing value check karne ka liye
df.isnull().sum()
```
