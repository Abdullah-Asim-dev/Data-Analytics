```python
import pandas as pd

df = pd.read_csv("Online_Retail_Cleaned.csv")
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
# Isse pata chalega sabse zyada revenue kis customer ne generate ki.
df.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10)
```

```python
# Isse pata chalega kin countries se sabse zyada revenue aayi.
df.groupby("CustomerID")["Revenue"].sum().sort_values(ascending=False).head(10)
```

```python
# Isse month-wise revenue milegi.
df.groupby("Country")["Revenue"].sum().sort_values(ascending=False).head(10)
```

```python
# Isse pata chalega kis din revenue sabse zyada hui.
df.groupby(df["InvoiceDate"].dt.to_period("M"))["Revenue"].sum()
```

```python
# datasetka overall revenue
df["Revenue"].sum()
```
