```python
import pandas as pd
```

```python
df = pd.read_csv("Online_Retail_Cleaned.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df.dtypes
```

```python
# ab hum na aik column bannaye ga jis ma hum Revenus nikale ga
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
# ab is ki head show karaye ga
df[["Quantity", "UnitPrice", "Revenue"]].head()
```

```python
# ab poore dataset ka revenus nikale ga
df["Revenue"].sum()
# # ab poore dataset ka revenus ka mean, mode, median
df["Revenue"].describe()
```

```python
# poore dataset ka revenue nikalo
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

```python
# average transaction revenue:
average_revenue = df["Revenue"].mean()

print("Average Revenue:", average_revenue)
```

```python
max_revenue = df["Revenue"].max()

print("Maximum Revenue:", max_revenue)
```

```python
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")
```

```python
df[["InvoiceDate", "YearMonth"]].head()
```

```python
df["DayName"] = df["InvoiceDate"].dt.day_name()
df[["InvoiceDate", "DayName"]].head()
```

```python
# Invoice Total
# .Invoice Total banate hain. Isse pata chalega ke ek complete invoice/order ki total sale kitni thi.
invoice_totals = df.groupby("InvoiceNo")["Revenue"].sum()
invoice_totals.head()
```

```python
# Customer Total
# Isse pata chalega ke har customer ne total kitni revenue generate ki.
customer_totals = df.groupby("CustomerID")["Revenue"].sum()
customer_totals.head()
```
