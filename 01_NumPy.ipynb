```python
# array creation
import numpy as np
arr = np.array([10, 20, 30, 40, 50])

print(arr)
```

```python
# array indexing
arr = np.array([10, 20, 30, 40, 50])
print(arr[0])
print(arr[1])
print(arr[2])
print(arr[4])
# negative indexing 
print(arr[-1])
print(arr[-2])
```

```python
arr=np.array([10,20,30,40,50])
print(arr[3])
print(arr[-1])
```

```python
# Array Slicing
# start sa value
arr=np.array([10,20,30,40,50])
print(arr[:3])
# end sa value
print(arr[2:])
# negative slice
print(arr[-3:])
# type checking
print(arr.dtype)
```

```python
arr_float = np.array([10.5, 20.5, 30.5])
# data type check
print(arr_float)
print(arr_float.dtype)
```

```python
# Copy vs View
# arr ki copy create karke us ma value chnage karna
# Independent array banata hai.
arr = np.array([10, 20, 30, 40, 50])

copy_arr = arr.copy()

copy_arr[0] = 100

print("Original:", arr)
print("Copy:", copy_arr)
```

```python
# view() original array ka view/reference banata hai. Isliye view mein change karoge to original bhi change ho jayega.
arr = np.array([10, 20, 30, 40, 50])

view_arr = arr.view()

view_arr[0] = 100

print("Original:", arr)
print("View:", view_arr)
```

```python
# Array Iterating
# Array ke har element ko one-by-one access karna iteration hai.
arr = np.array([10, 20, 30, 40, 50])

for value in arr:
    print(value)
```

```python
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

for row in arr_2d:
    print(row)
```

```python
# Agar har individual element print karna ho:
for row in arr_2d:
    for value in row:
        print(value)
```

```python
# NumPy multidimensional arrays ko iterate karne ke liye nditer() bhi provide karta hai:
for value in np.nditer(arr_2d):
    print(value)
```

```python
# Array Join
# Do arrays ko combine karne ke liye np.concatenate()
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])

joined = np.concatenate((arr1, arr2))

print(joined)
```

```python
# axis=0 — rows ke neeche rows
arr1 = np.array([
    [10, 20],
    [30, 40]
])

arr2 = np.array([
    [50, 60],
    [70, 80]
])

joined = np.concatenate((arr1, arr2), axis=0)

print(joined)
```

```python
# axis=1— rows ke neeche column
arr1 = np.array([
    [10, 20],
    [30, 40]
])

arr2 = np.array([
    [50, 60],
    [70, 80]
])

joined = np.concatenate((arr1, arr2), axis=1)

print(joined)
```

```python
# Array Split
arr = np.array([10, 20, 30, 40, 50, 60])
split_arr = np.array_split(arr, 3)
# daary divide in parts
print(split_arr)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

split_arr = np.array_split(arr, 2)

print(split_arr)
```

```python
# Array Search
# . where() — value ka index find karna
arr = np.array([10, 20, 30, 40, 50, 30])

result = np.where(arr == 30)

print(result)
```

```python
result = np.where(arr > 20)

print(result)
```

```python
arr = np.array([10, 20, 30, 40, 50])
# value search in sorted array
result = np.searchsorted(arr, 35)

print(result)
```

```python
arr = np.array([10, 20, 30, 40, 50, 30])

print(np.where(arr == 30))
print(np.where(arr > 20))
```

```python
arr = np.array([50, 10, 40, 20, 30])
# arr ko sort karna
sorted_arr = np.sort(arr)

print(sorted_arr)
```

```python
print("Original:", arr)
print("Sorted:", sorted_arr)
```

```python
arr_2d = np.array([
    [30, 10, 20],
    [60, 40, 50]
])

print(np.sort(arr_2d))
```

```python
# Array Filtering
# Filtering ka matlab hai condition ke basis par array se specific values nikalna.
arr = np.array([10, 20, 30, 40, 50])
filter_arr = arr > 30

print(filter_arr)
```

```python
new_arr = arr[arr >= 20]

print(new_arr)
```

```python
new_arr = arr[arr % 2 == 0]

print(new_arr)
```
